```python
file_path = "/kaggle/input/sleep-health-and-lifestyle-dataset/Sleep_health_and_lifestyle_dataset.csv"
df = pd.read_csv(file_path)
```

```python
init_notebook_mode(connected=True)
```

```python
query = """
    SELECT `Occupation`, `Sleep Duration`, `Quality of Sleep`
    FROM df;
"""

# Execute the query
sleep_by_occupation = sqldf(query, globals())  # 'globals()' makes 'df' available in SQL
display(sleep_by_occupation) # Display results
```

```python
import plotly.express as px

fig = px.scatter(
    sleep_by_occupation,
    x='Sleep Duration',
    y='Quality of Sleep',
    color='Occupation',
    title='Sleep Duration vs Quality of Sleep by Occupation',
    labels={
        'Sleep Duration': 'Sleep Duration (hours)',
        'Quality of Sleep': 'Quality of Sleep (1-10)'
    },
    height=600
)

fig.show(renderer='iframe')
```

```python
query = """
    select `occupation`,`Physical Activity Level` as pal, `stress level`
    FROM df;
"""

# Execute the query
sleep_by_stress = sqldf(query, globals())  # 'globals()' makes 'df' available in SQL
display(sleep_by_stress)  # Display results
```

```python
fig2 = px.bar(sleep_by_stress, 
              x='Occupation', 
              y='pal', 
              color='Stress Level',
              title='Physical Activity Level by Occupation with Stress Level',
              labels={'pal': 'Physical Activity Level (minutes/day)'},
              barmode='group')
fig2.update_layout(height=600, xaxis={'categoryorder':'total descending'})
fig2.show(renderer='iframe')
```

```python
query= """
select Occupation, `Sleep Disorder`
from df;
"""
quality_of_sleep = sqldf(query, globals())
display(quality_of_sleep)  # Display results
```

```python
query= """
select `BMI Category`, `Blood Pressure`, `Sleep Duration`
from df;
"""
result4 = sqldf(query, globals())
display(result4)  # Display results
```

```python
fig4 = px.box(
    result4,
    x="BMI Category",
    y="Sleep Duration",
    color="Blood Pressure",  # Only works if Blood Pressure is categorical
    title="Sleep Duration by BMI Category & Blood Pressure",
    labels={"Sleep Duration": "Sleep Duration (hours)"},
)
fig4.update_layout(height=600)
fig4.show(renderer="iframe")
```

```python
query = """
SELECT 
    `Occupation`, 
    AVG(`Stress Level`) AS 'Stress Level',
    AVG(`Blood Pressure`) AS 'Avg Blood Pressure',
    COUNT(*) AS 'Count'
FROM df
GROUP BY `Occupation`
"""
result5 = sqldf(query, globals())
display(result5)
```

```python
fig5 = px.scatter(
    result5,
    x='Stress Level',
    y='Avg Blood Pressure',
    size='Count',
    color='Occupation',
    title='Stress Level vs Blood Pressure by Occupation',
    labels={
        'Avg Blood Pressure': 'Avg Blood Pressure (mmHg)',
        'Stress Level': 'Stress Level (1-10)'
    },
    hover_data=['Occupation', 'Avg Blood Pressure', 'Stress Level', 'Count']
)
fig5.update_layout(height=600)
fig5.show(renderer='iframe')
```
